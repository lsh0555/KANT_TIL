### chapter13 - 딕셔너리

딕셔너리(Dictionary) : 데이터를 `키(key) : 값(value)` 형태로 저장하는 자료구조

student = {
    "name": "민수",
    "age": 20,
    "score": 85
}

* `"name"`, `"age"`, `"score"` → key
* `"민수"`, `20`, `85` → value

- 리스트는 인덱스(위치)로 값을 찾지만, 딕셔너리는 key를 이용해 값을 찾음
- 딕셔너리에 존재하지 않는 key를 `[]`로 조회하면 `KeyError` 발생
- 한 대상의 여러 정보를 표현하기 좋음

data = {}
 + `{}`는 빈 집합이 아니라 빈 딕셔너리


< 딕셔너리 값 추가와 수정 >
딕셔너리는 추가와 수정에 같은 대입 문법 사용

student["city"] = "서울"

`city`가 기존에 없다면 새로운 데이터가 추가됨
이미 `city`가 있다면 기존 값이 변경됨

< 딕셔너리 값 삭제 >

 * `del` : 특정 key와 value 삭제

del student["city"]
 * `pop()` : 항목을 삭제하면서 삭제된 value 반환

removed = student.pop("score")



< 딕셔너리 값 조회 방법>>

* `[]` : 반드시 존재해야 하는 key를 조회할 때 사용

student["name"]

* `get()` : key가 없을 수도 있을 때 안전하게 조회

student.get("city")

 1) key가 없다면 기본적으로 `None` 반환
 2) 기본값을 직접 지정 가능
     ㄴ student.get("city", "정보 없음")

* `in` : key가 존재하는지 먼저 확인

if "score" in student:
    print(student["score"])


< 딕셔너리 key, value를 확인 >
* `keys()` : key만 가져옴
student.keys()

* `values()` : value만 가져옴
student.values()

* `items()` : key와 value를 한 쌍으로 가져옴
student.items()

< 딕셔너리 반복문 >
딕셔너리도 for문 사용 가능
1) key만 반복

for key in student:
    print(key)

2) key, value 함께 반복

for key, value in student.items():
    print(key, value)

+ `items()`는 `(key, value)` 한 쌍을 가져오기 때문에 두 개의 변수로 받을 수 있음

< 딕셔너리 리스트>
여러 명의 학생을 저장하고 싶다면 딕셔너리를 리스트에 사용

students = [
    {"name": "민수", "score": 85},
    {"name": "지영", "score": 92},
    {"name": "서준", "score": 78}
]

< 딕셔너리 반복문 >
`student`에는 반복할 때마다 학생 한 명의 딕셔너리

for student in students:
    print(student["name"], student["score"])


< 중첩 딕셔너리 >
딕셔너리의 value에 또 다른 딕셔너리가 들어갈 수 있음
바깥쪽부터 안쪽으로 한 단계씩 접근

student = {
    "name": "민수",
    "contact": {
        "email": "minsu@example.com",
        "phone": "010-1234-5678"
    }
}

email을 가져오려면:
student["contact"]["email"]


 + 오류 확인

복잡한 딕셔너리에서 오류가 발생하면 한 번에 접근하기보다 단계별로 확인하는 것이 좋다.

print(student.keys())
print(student["contact"])
print(student["contact"].keys())

이렇게 확인하면 어떤 단계에서 원하는 key가 없는지 찾기 쉽다.

```python
# mission 01 - 상품 정보 안전하게 조회하기
product = {
    "name": "키보드",
    "price": 39000
}

print(product["name"])
print(product["price"])
product.get("stock")
print("stock", "재고 정보 없음")
"price" in product
if "price" in product:
    print(product["price"])
```

```python
# mission 02 - 80점 이상 학생만 출력하기
students = [
    {"name": "민수", "score": 85},
    {"name": "지영", "score": 92},
    {"name": "서준", "score": 78}
]
for student in students:
    if student["score"] >= 80:
        print(student["name"], student["score"])
```

```python
# mission 03 - 상품 목록 관리하기
products = [
    {"name": "사과", "price": 50000, "stock": 10},
    {"name": "바나나", "price": 20000, "stock": 3},
    {"name": "딸기", "price": 80000, "stock": 0}
]

print(products)
# 가격 30000원 이상의 상품 출력
for product in products:
    if product["price"] >= 30000:
        print(product["name"])

# stock이 0인 상품을 찾아 품절 출력
for product in products:
    if product["stock"] == 0:
        print(f"{product["name"]} 상품은 품절입니다.")
```
