### chapter16 - 파일과 CSV 데이터 다루기
 - 파일을 읽고 저장하는 방법과 CSV 데이터를 처리하는 방법

1. 파일 읽기와 쓰기 : `with open()`
  ㄴ 파일을 열어서 내용을 읽거나 새로운 내용 저장 가능
     `with`를 사용하면 작업이 끝난 후 파일이 자동으로 닫히기 때문에 안전하게 파일 관리 가능

[ 파일을 열 때 사용하는 모드 ]

* `r` : 파일 읽기
* `w` : 파일 쓰기, 기존 내용이 있으면 덮어쓰기
* `a` : 기존 내용 뒤에 새로운 내용 추가하기


2. 파일을 읽는 방법
 1) 파일 전체 한 번에 읽을 때 : `read()`
 2) 한 줄만 읽을 때 : `readline()`

 + 파일 내용 한 줄씩 순서대로 처리 : `for` 반복문 사용

3. 파일 경로

파일 이름이 정확하더라도 파일의 실제 위치와 코드에서 지정한 경로가 다르면 `FileNotFoundError` 발생 가능
* 파일 이름뿐만 아니라 파일이 저장되어 있는 폴더와 경로 확인도 필수

`pathlib.Path`를 사용하면 프로젝트를 기준으로 파일 경로 구성 가능.

### 4. CSV(Comma-Separated Values) 파일
  데이터를 쉼표와 줄바꿈으로 구분하여 표 형태로 저장하는 텍스트 파일
  Python의 `csv` 모듈을 이용하여 CSV 데이터를 읽고 저장 가능
  `DictReader`를 사용하면 `row["price"]`처럼 열 이름으로 값을 가져올 수 있어서 데이터의 의미 파악 쉬움

* `csv.reader` : 데이터를 리스트 형태로 읽기
* `csv.DictReader` : 열 이름을 이용하여 딕셔너리 형태로 읽기
* `csv.writer` : CSV 파일에 데이터 저장하기

### 5. CSV 데이터 계산
CSV에서 읽어온 숫자는 기본적으로 문자열이라 바로 숫자 계산에 사용 불가

`int()` 또는 `float()`로 숫자로 변환한 후 계산해야 함

  파일 읽기 → 한 행 가져오기 → 필요한 열 선택 → 숫자 변환 → 계산 → 결과 누적


### 6. CSV 파일 저장
`csv.writer`를 이용하면 새로운 CSV 파일 생성

* `writerow()` : 한 행 저장
* `writerows()` : 여러 행 저장

## 오류가 발생했을 때 확인할 것
파일이나 CSV를 처리할 때 오류가 발생하면 코드만 보는 것이 아니라 파일의 실제 상태도 함께 확인 필요

* `FileNotFoundError` → 파일 이름, 확장자, 폴더 위치, 실제 경로 확인
* 한글 깨짐 → 인코딩 확인
* `KeyError` → CSV의 헤더 이름 확인
* `ValueError` → 숫자로 변환하려는 실제 데이터 확인

### chapter17 - 오류와 예외 처리
Python 프로그램을 실행하면서 발생할 수 있는 오류의 종류와 예외를 처리하는 방법

### 1. 오류의 종류
Python에서 발생하는 문제는 크게 세 가지로 구분 가능

* 문법 오류 : Python 문법 자체가 잘못된 경우
* 실행 중 예외 : 문법은 맞지만 프로그램 실행 중 문제가 발생한 경우
* 논리 오류 : 프로그램은 정상적으로 실행되지만 내가 원하는 결과와 다른 경우

오류가 발생했을 때 바로 코드를 수정하기보다 어떤 종류의 오류인지 먼저 파악하는 것이 중요

### 2. 트레이스백 확인
프로그램에서 오류가 발생하면 트레이스백을 통해 오류의 원인 가능

[ 트레이스백 확인 순서 ex. ]

1. 마지막 줄에서 예외 이름 확인         // 오류 메시지가 길더라도 마지막 줄부터 확인하면 어떤 오류인지 조금 더 쉽게 파악 가능
2. 오류가 발생한 파일과 줄 번호 확인
3. 해당 줄에서 실행된 코드 확인
4. 코드에 들어간 실제 값 확인

### 3. try-except 
 ㄴ 실행 중 발생할 것으로 예상되는 예외 처리 가능
    `try` -> 문제가 발생할 가능성이 있는 코드 작성, `except`-> 해당 문제가 발생했을 때 실행할 코드 작성

### 4. 자주 발생하는 예외

* `ValueError` : 잘못된 형식의 값을 변환할 때
* `ZeroDivisionError` : 숫자를 0으로 나눌 때
* `FileNotFoundError` : 존재하지 않는 파일을 열려고 할 때
* `KeyError` : 딕셔너리에 존재하지 않는 키를 사용할 때
* `IndexError` : 리스트의 범위를 벗어난 위치에 접근할 때

여러 문제가 발생할 수 있는 코드에서는 예외 종류에 따라 `except`를 각각 작성하여 다르게 처리 가능

### 5. else와 finally

* `else` : `try`에서 예외가 발생하지 않았을 때 실행
* `finally` : 성공하거나 실패하는 것과 관계없이 마지막에 실행

### 6. 예외 처리와 값 검증의 차이
   ㄴ 예외 처리와 데이터 검증은 서로 다른 문제

점수에 `"abc"`를 입력 -> `ValueError` 발생
점수의 범위가 0~100일때 `150`은 예외 처리가 아니라 조건문을 이용한 값의 범위 검증 필요

### 7. 올바른 값이 입력될 때까지 반복하기 - `while` 반복문 사용
   ㄴ 사용자가 올바른 값을 입력할 때까지 다시 입력받을 수 있다.

  [ 입력 → 형 변환 → 예외 처리 → 값의 범위 확인 → 정상 값이면 반복 종료 ]

### 8. 모든 오류를 무조건 처리하면 안 된다

`except:`처럼 예외 이름을 작성하지 않고 모든 오류를 처리하면 어떤 문제가 발생했는지 알기 어려워질 수 있다.

예상하지 못한 개발 오류까지 숨길 수 있기 때문에 가능하면 `ValueError`, `ZeroDivisionError`처럼 예상되는 예외 이름을 구체적으로 작성하는 것이 좋다.
