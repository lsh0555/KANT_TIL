## Chapter03 - PostgreSQL과 DBeaver로 데이터베이스 실습 환경 만들기

1. PostgreSQL과 DBeaver의 차이

PostgreSQL: 실제 데이터를 저장하고 SQL을 실행하는 관계형 DBMS

DBeaver: PostgreSQL에 접속하여 SQL을 작성하고 실행 결과를 확인하는 데이터베이스 클라이언트

사용자
  ↓
DBeaver에서 SQL 작성
  ↓
PostgreSQL에 SQL 전달
  ↓
PostgreSQL이 SQL 실행
  ↓
DBeaver에서 결과 또는 오류 확인

따라서 DBeaver를 종료하더라도 PostgreSQL에 저장된 데이터가 삭제되는 것은
아니다.

2. 데이터베이스 연결 정보

항목       의미                         로컬 환경 예시

Host       PostgreSQL 서버의 주소       localhost
Port       서버에 접속하는 통로         5432
Database   접속할 데이터베이스          postgres
Username   로그인할 PostgreSQL 사용자   postgres
Password   해당 사용자의 인증 정보      설치 시 설정한 값

postgres는 사용자 이름과 데이터베이스 이름으로 모두 사용될 수 있으므로
의미를 구분해야 한다.

3. 데이터베이스 생성과 연결은 다르다

새 데이터베이스를 생성해도 현재 연결 위치가 자동으로 변경되는 것은
아니다.

데이터베이스 생성 ≠ 현재 연결 대상 변경

현재 연결된 데이터베이스는 다음 SQL로 확인할 수 있다.

SELECT current_database();

4. 현재 데이터베이스와 스키마 확인

SELECT current_database();
SELECT current_user;
SELECT current_schema();
SHOW search_path;

current_database() : 현재 연결된 데이터베이스

current_user : 현재 접속한 PostgreSQL 사용자

current_schema() : 현재 사용되는 스키마

SHOW search_path : 스키마 이름을 생략했을 때 객체를 찾는 순서

5. SQL 실행 범위 구분

현재 문장 실행 → 현재 SQL 한 문장 실행
선택 영역 실행 → 선택한 SQL만 실행
전체 스크립트 실행 → 여러 SQL 문장을 실행

SQL 실행 전 확인할 것:

어느 데이터베이스에 연결되어 있는가?

어떤 SQL을 실행하려고 하는가?

Auto-commit 상태는 어떻게 되어 있는가?

6. 데이터베이스가 보이지 않을 때

1. 현재 연결된 Database 확인
2. Schema 확인
3. DBeaver 필터 확인
4. 새로고침(Refresh)

데이터베이스나 테이블이 보이지 않는다고 바로 생성 실패라고 판단하지 않고
현재 연결 위치와 화면 상태를 먼저 확인해야 한다.

7. 비밀번호와 접속 정보 관리

실제 비밀번호를 SQL 파일에 작성하지 않는다.

전체 데이터베이스 접속 URL을 공개하지 않는다.

화면 캡처 전에 개인정보와 연결 정보를 확인한다.

비밀번호나 API Key가 노출되었다면 변경하거나 폐기한다.

💡 새롭게 알게 된 점

PostgreSQL은 데이터를 저장하고 SQL을 실행하는 DBMS이고, DBeaver는
PostgreSQL에 연결해 SQL을 작성하고 결과를 확인하는 클라이언트라는
것을 알게 되었다.

데이터베이스를 새로 생성했다고 현재 연결 위치가 자동으로 새
데이터베이스로 변경되는 것은 아니라는 것을 알게 되었다.

SQL 실행 전에 현재 데이터베이스, 사용자, 스키마와 실행 범위를
확인하는 습관이 중요하다는 것을 알게 되었다.
