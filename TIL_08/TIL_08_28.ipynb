## 2. `__init__()` : 객체를 생성할 때 처음 필요한 값을 저장하기 위해 사용하는 메서드
객체가 생성될 때 초기 상태를 준비하는 역할

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

-> 객체를 생성하면서 값을 전달할 수 있다.

student1 = Student("민수", 85)
student2 = Student("지영", 92)

-> 각 객체는 서로 다른 값을 저장한다.

print(student1.name)
print(student1.score)

print(student2.name)
print(student2.score)

결과:

민수
85
지영
92

## 4. 속성(Attribute) : 객체가 가지고 있는 데이터

속성값을 수정할 때, 클래스 자체가 변경되는 것이 아니라 해당 객체의 속성값이 변경됨

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock
```

여기서

```python
self.name
self.price
self.stock
```

이 객체의 속성이다.

객체를 만든 후 속성을 읽을 수 있다.

```python
product = Product("키보드", 45000, 10)

print(product.name)
print(product.price)
print(product.stock)
```

속성값을 수정하는 것도 가능하다.


product.price = 42000
product.stock = 8


## 5. 메서드(Method) : 클래스 안에 정의한 함수

함수와 메서드 둘 다 기능을 묶는다는 점은 같지만, 메서드는 객체와 연결되어 `self`를 통해 객체의 데이터 사용 가능

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def show_result(self):
        print(f"{self.name}: {self.score}점")

    def is_passed(self):
        return self.score >= 60
```

객체를 통해 메서드를 호출할 수 있다.

```python
student = Student("민수", 85)

student.show_result()
print(student.is_passed())
```

결과:

```text
민수: 85점
True
```


# 6. 하나의 클래스로 여러 객체 만들기

하나의 클래스에서 여러 개의 객체를 만들 수 있다.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def show_info(self):
        print(f"{self.name}: {self.price:,}원")
```

```python
keyboard = Product("키보드", 45000)
mouse = Product("마우스", 25000)
webcam = Product("웹캠", 65000)
```

같은 `Product` 클래스를 사용하지만 각각 다른 데이터를 가진 객체이다.

```text
keyboard → 키보드 / 45000
mouse    → 마우스 / 25000
webcam   → 웹캠 / 65000
```

# 7. 객체를 리스트에 저장하기

객체도 리스트의 요소로 저장할 수 있다.

```python
products = [
    Product("키보드", 45000),
    Product("마우스", 25000),
    Product("웹캠", 65000)
]
```

반복문을 사용하면 여러 객체를 한 번에 처리할 수 있다.

```python
for product in products:
    product.show_info()
```

객체의 속성을 이용한 계산도 가능하다.

```python
total = 0

for product in products:
    total += product.price

print(total)
```

### 연결해서 생각하기

```text
리스트
 ↓
객체 여러 개 저장
 ↓
for 반복문
 ↓
객체 하나씩 꺼내기
 ↓
속성 또는 메서드 사용
```

# 9. 클래스에서 자주 발생하는 오류

클래스에서 오류가 발생하면 다음 순서로 확인한다.

1. 클래스 이름과 객체 생성 코드
2. `__init__()`의 매개변수
3. 객체를 생성할 때 전달한 값의 개수
4. 메서드에 `self`가 있는지
5. 실제 생성한 속성 이름과 사용한 속성 이름
6. `TypeError`인지 `AttributeError`인지 확인

---

## TypeError

객체 생성에 필요한 값을 제대로 전달하지 않았을 때 발생할 수 있다.

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
```

잘못된 코드:

```python
student = Student("민수")
```

`name`, `score` 두 개의 값이 필요한데 `"민수"` 하나만 전달했기 때문에 문제가 발생한다.

올바른 형태:

```python
student = Student("민수", 85)
```

`self`는 직접 전달하지 않는다는 점도 중요하다.

---

## AttributeError

객체에 존재하지 않는 속성을 사용할 때 발생할 수 있다.

```python
print(student.grade)
```

그런데 실제 클래스에는

```python
self.name = name
self.score = score
```

만 존재한다면 `grade`라는 속성이 없기 때문에 오류가 발생한다.

따라서 `AttributeError`가 발생하면 **`__init__()`에서 실제로 만든 속성 이름을 먼저 확인한다.**

# 11. 종합 예제 - 상품 재고 관리

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def sell(self, quantity):
        if quantity <= self.stock:
            self.stock -= quantity
            return True

        return False

    def show_info(self):
        print(f"{self.name}: {self.price:,}원 / 재고 {self.stock}개")
```

객체 생성:

```python
product = Product("키보드", 45000, 10)
```

현재 정보 확인:

```python
product.show_info()
```

상품 3개 판매:

```python
print(product.sell(3))
```

다시 정보를 확인하면 재고가 감소한다.

```python
product.show_info()
```

흐름:

```text
객체 생성
 ↓
stock = 10
 ↓
sell(3)
 ↓
self.stock -= 3
 ↓
stock = 7
```

즉, **메서드를 통해 객체 자신의 상태를 변경할 수도 있다.**
