### chapter 11 - 리스트

리스트(List) : 여러 개의 값을 하나의 변수에 순서대로 저장하는 자료구조

[ 리스트의 특징 ]
* `[]` 대괄호 사용
* 여러 개의 값 하나의 변수에 저장 가능
* 값마다 순서(index) 존재
* 중복된 값 저장 가능
* 생성한 이후에도 값 추가, 수정, 삭제 가능


fruits = ["사과", "바나나", "포도"]

| 값      | 사과 | 바나나 | 포도 |
| ------ | -- | --- | -- |
| 인덱스    | 0  | 1   | 2  |
| 음수 인덱스 | -3 | -2  | -1 |

* 첫 번째 인덱스는 `0`부터 시작
* `-1`을 사용하면 마지막 값을 가져옴
* 존재하지 않는 위치에 접근하면 `IndexError' 발생


슬라이싱(Slicing) : 리스트에서 특정 범위의 여러 값을 가져오는 방법

numbers[start:end]
** end 위치는 포함하지 않는다


< 리스트 값 추가 >

`append()` : 리스트의 맨 뒤에 값 하나 추가
`insert()` : 리스트에서 원하는 특정 위치에 값 하나 추가
`extend()` : 리스트에 다른 리스트의 여러 값 이어 붙이기


< 리스트 값 삭제 >
`remove(value)` : 리스트에서 특정 값을 찾아서 삭제
  + 없는 값을 삭제하려고 하면 `ValueError` 발생

`pop(index)` : 리스트에서 특정 위치의 값을 꺼내면서 삭제
 + 삭제된 값을 변수에 저장 가능

`del` : 리스트의 특정 위치의 값을 바로 삭제

< 리스트 상태 확인 >

`len()` : 리스트에 저장된 값의 개수 확인

`in` : 리스트에 특정 값이 존재하는지 확인
 + 결과는 `True` 또는 `False`이다.

`count()` : 리스트에 특정 값이 몇 번 들어 있는지 확인

`index()` : 특정 값이 처음 등장하는 위치 확인
 + 찾는 값이 없으면 `ValueError` 발생


< 리스트 정렬 >

`sort()` : 리스트 자체를 오름차순으로 정렬
`reverse()` : 현재 리스트의 순서를 반대로 뒤집음

### chapter 12 튜플, 집합

| 자료구조    | 순서 | 수정 | 중복 | 인덱싱 |
| ------- | -- | -- | -- | --- |
| `list`  | O  | O  | O  | O   |
| `tuple` | O  | X  | O  | O   |
| `set`   | X  | O  | X  | X   |

튜플(Tuple) : 여러 개의 값을 순서대로 저장하는 자료구조
 + 리스트처럼 여러 데이터를 저장할 수 있지만, 튜플은 요소를 직접 수정할 수 없음
 + 소괄호 () 사용
 + 둘 다 순서가 있기 때문에 인덱싱과 슬라이싱 가능
 + 요소가 하나일 때 반드시 쉼표를 작성해야 함 / 쉼표 사용 안 하면 그냥 정수(int)가 됨
 ㄴ one_value = (10,)

< 튜플 패킹과 언패킹 >
패킹(Packing) : 여러 개의 값을 하나의 튜플로 묶는 것
언패킹(Unpacking) : 튜플에 들어 있는 여러 값을 각각의 변수로 나누어 저장하는 것
 ** 변수의 개수와 튜플의 값 개수가 일치해야 함

집합(Set) : 중복을 허용하지 않는 자료구조
 ( 같은 값을 여러 번 넣어도 하나만 저장됨 -> 중복 제거 및 고유한 데이터 관리에 유용 )
 + 집합은 인덱싱할 수 없음
 + 리스트나 튜플처럼 순서에 따라 데이터 관리하지 않음 -> 인덱싱 사용 불가능
 + 특정 위치의 값을 가져오는 것보다 데이터의 존재 여부나 집합 간 관계를 확인하는 데 적합
 + 집합의 출력 순서를 리스트처럼 특정 순서로 기대해서도 안 됨

< 빈 집합 만들기 >
empty_set = set()

{}를 사용하면 빈 집합이 아니라 빈 딕셔너리가 되기에 주의해야 함

empty_set = set()
empty_dict = {}

print(type(empty_set))
-> <class 'set'>

print(type(empty_dict))
-> <class 'dict'>

빈 집합 → set()
빈 딕셔너리 → {}


< 집합에 값 추가하기 >
집합에 값을 추가할 때 add() 사용 -> 이미 존재하는 값 추가해도 중복으로 저장 안 됨

skills = {"Python", "SQL"}
skills.add("AI")

< 집합에서 값 삭제하기 >
존재하는 값을 삭제하면 remove()가 실행되지만 존재하지 않는 값을 삭제하면 KeyError 발생
remove()
skills.remove("SQL")

존재하는 값을 삭제하면 discard()가 실행되며, 만약 존재하지 않아도 오류 발생하지 않음
skills.remove("Java")
discard()
skills.discard("Java")

< 집합 연산 >
집합끼리 비교 연산이 가능함

1. 합집합 | : 두 집합의 모든 값을 합침
A = {1, 2, 3, 4}
B = {3, 4, 5}

결과: {1, 2, 3, 4, 5}

2. 교집합 & : 두 집합에 공통으로 존재하는 값을 찾음
A & B

결과: {3, 4}

A와 B 모두에 존재하는 값

3. 차집합 - : 첫 번째 집합에는 있지만 두 번째 집합에는 없는 값을 찾음 (어느 집합을 먼저 작성하는지 중요)
A - B

결과: {1, 2}

반대로: B - A
결과: {5}

```python
# mission 02 - 기술 집합 관리하기
skills = {"Python", "SQL"}
skills.add("AI")
skills.add("Python")  #  집합set은 중복을 제거하기 때문에 Python 부분은 추가 되지 않는다.

skills.remove("SQL")
skills.discard("Java")  # 에러 없이 그냥 진행 된다.  // remove()였으면 KeyError 발생

print(skills)
```

```python
# 종합 실습 - 두 팀의 기술 목록 비교하기
team_a_list = ["Python", "SQL", "AI", "Python"]
team_b_list = ["Python", "Java", "AI", "Java"]

team_a = []
team_b = []

# 중복 제거하여 team_a에 넣기
for a in team_a_list:
    if a not in team_a:
        team_a.append(a)

for b in team_b_list:
    if b not in team_b:
        team_b.append(b)


print("A팀 고유 기술:", team_a)
print("B팀 고유 기술:", team_b)
# set에서는 집합 연산자 (&, |, - ) 사용 가능
# list로도 중복을 없앨 수 있지만 이 경우 team_a/ team_b를 set으로 저장하는게 코드 간결해짐
print("공통 기술:", set(team_a) & set(team_b)) # 교집합
print("전체 기술:", set(team_a) | set(team_b)) # 합집합
print("A팀만:", set(team_a) - set(team_b))   # A에만 있는 차집합 
print("B팀만:", set(team_b) - set(team_a))   # B에만 있는 차집합
```

```python
# 심화 과제 - set 사용하지 않고 list 기능만 사용하여 unique_tags 구현하기
tags = ["Python", "AI", "Python", "Data", "AI"]
print(tags)

unique_tags = []

for tag in tags: 
    if tag not in unique_tags:
        unique_tags.append(tag)

print(unique_tags)
```
