### chapter 11 - 리스트

리스트(List) : 여러 개의 값을 하나의 변수에 순서대로 저장하는 자료구조

[ 리스트의 특징 ]
* `[]` 대괄호 사용
* 여러 개의 값 하나의 변수에 저장 가능
* 값마다 순서(index) 존재
* 중복된 값 저장 가능
* 생성한 이후에도 값 추가, 수정, 삭제 가능


fruits = ["사과", "바나나", "포도"]

| 값      | 사과 | 바나나 | 포도 |
| ------ | -- | --- | -- |
| 인덱스    | 0  | 1   | 2  |
| 음수 인덱스 | -3 | -2  | -1 |

* 첫 번째 인덱스는 `0`부터 시작
* `-1`을 사용하면 마지막 값을 가져옴
* 존재하지 않는 위치에 접근하면 `IndexError' 발생


슬라이싱(Slicing) : 리스트에서 특정 범위의 여러 값을 가져오는 방법

numbers[start:end]
** end 위치는 포함하지 않는다


< 리스트 값 추가 >

`append()` : 리스트의 맨 뒤에 값 하나 추가
`insert()` : 리스트에서 원하는 특정 위치에 값 하나 추가
`extend()` : 리스트에 다른 리스트의 여러 값 이어 붙이기


< 리스트 값 삭제 >
`remove(value)` : 리스트에서 특정 값을 찾아서 삭제
  + 없는 값을 삭제하려고 하면 `ValueError` 발생

`pop(index)` : 리스트에서 특정 위치의 값을 꺼내면서 삭제
 + 삭제된 값을 변수에 저장 가능

`del` : 리스트의 특정 위치의 값을 바로 삭제

< 리스트 상태 확인 >

`len()` : 리스트에 저장된 값의 개수 확인

`in` : 리스트에 특정 값이 존재하는지 확인
 + 결과는 `True` 또는 `False`이다.

`count()` : 리스트에 특정 값이 몇 번 들어 있는지 확인

`index()` : 특정 값이 처음 등장하는 위치 확인
 + 찾는 값이 없으면 `ValueError`가 발생할 수 있다.

## 8. 리스트 + 반복문

`for`문을 사용하면 리스트의 값을 **처음부터 하나씩 꺼내서 반복 처리**할 수 있다.

```python
scores = [85, 92, 78]

for score in scores:
    print(score)
```

실행 흐름:

```text
85 → score에 저장 → 실행
92 → score에 저장 → 실행
78 → score에 저장 → 실행
종료
```

리스트와 `for`문을 함께 사용하면 데이터가 많아져도 같은 작업을 반복해서 처리할 수 있다.

---

## 9. 리스트 + 반복문 + 조건문

리스트의 모든 값을 확인하면서 특정 조건에 맞는 값만 처리할 수도 있다.

```python
for score in scores:
    if score >= 80:
        print(score)
```

즉,

**리스트 → 여러 데이터 저장**
**for → 데이터를 하나씩 확인**
**if → 조건에 맞는 데이터 선택**

의 관계로 이해할 수 있다.

## 10. 숫자 리스트 집계

숫자가 저장된 리스트는 다음 함수로 간단하게 계산할 수 있다.

| 함수      | 기능    |
| ------- | ----- |
| `sum()` | 합계    |
| `min()` | 최솟값   |
| `max()` | 최댓값   |
| `len()` | 값의 개수 |

평균은 다음 원리로 계산한다.

```python
평균 = 합계 / 개수
```

즉 `sum()`과 `len()`을 함께 사용할 수 있다.

## 11. 리스트 정렬

### `sort()`

리스트 자체를 오름차순으로 정렬한다.

```python
numbers.sort()
```

```text
[30, 10, 40, 20]
        ↓
[10, 20, 30, 40]
```

### `reverse()`

현재 리스트의 순서를 반대로 뒤집는다.

```python
numbers.reverse()
```

`sort()`와 `reverse()`는 리스트 자체의 상태를 변경한다.

## 12. 자주 발생하는 오류

### `IndexError`

존재하지 않는 인덱스에 접근할 때 발생한다.

```python
items = [10, 20, 30]

items[3]  # 오류
```

길이가 3인 리스트의 인덱스는 `0, 1, 2`까지만 존재한다.

### `ValueError`

리스트에 존재하지 않는 값을 `remove()`하거나 `index()`로 찾을 때 발생할 수 있다.

```python
fruits.remove("없는 값")
```

따라서 값을 삭제하거나 찾기 전에 `in`을 이용해 존재 여부를 확인할 수 있다.
