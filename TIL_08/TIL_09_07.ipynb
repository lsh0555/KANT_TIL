## Chapter 04. 관계형 데이터베이스와 SQL 시작하기

1. CRUD와 SQL

CRUD : 데이터베이스에서 데이터를 다루는 기본 작업

| CRUD | SQL | 기능 |
|---|---|---|
| Create | `INSERT` | 새로운 데이터 추가 |
| Read | `SELECT` | 데이터 조회 |
| Update | `UPDATE` | 기존 데이터 수정 |
| Delete | `DELETE` | 데이터 삭제 |

### 주의할 점
- `CREATE TABLE` : 데이터를 저장할 테이블 구조 생성
- `INSERT` : 만들어진 테이블에 실제 데이터 추가

2. DBeaver에서 현재 연결 확인하기

SQL을 실행하기 전에 현재 어떤 데이터베이스에 연결되어 있는지 확인하는 것이 중요하다.

```sql
SELECT current_database();
SELECT current_user;
SELECT current_schema();
SHOW search_path;
```

### 각각의 의미

- `current_database()` : 현재 연결된 데이터베이스
- `current_user` : 현재 접속한 사용자
- `current_schema()` : 현재 기본 스키마
- `search_path` : PostgreSQL이 테이블 등을 찾는 스키마 순서

DBeaver 화면에 보이는 연결 이름만 믿는 것이 아니라 SQL을 직접 실행해서 실제 연결 상태를 확인하는 습관이 중요하다.

# 3. SQL 기본 작성 규칙

SQL을 작성할 때 다음과 같은 규칙을 사용한다.

- SQL 명령어는 대문자로 작성
- 테이블명과 컬럼명은 소문자로 작성
- 문자열은 작은따옴표 `' '` 사용
- 숫자는 따옴표 없이 작성
- SQL 문장 마지막에는 세미콜론 `;` 사용
- 한 줄 주석은 `--` 사용

예시:

```sql
-- 컴퓨터공학 학생 조회
SELECT *
FROM public.students
WHERE major = '컴퓨터공학';
```

# 5. CREATE TABLE - 테이블 생성

`CREATE TABLE`은 데이터를 저장할 테이블의 구조를 만드는 명령어이다.

```sql
CREATE TABLE public.students (
    id INTEGER GENERATED BY DEFAULT AS IDENTITY PRIMARY KEY,
    name VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE NOT NULL,
    major VARCHAR(100),
    grade INTEGER,
    created_at TIMESTAMPTZ NOT NULL DEFAULT CURRENT_TIMESTAMP
);
```

## 주요 제약조건

### PRIMARY KEY

각 데이터를 고유하게 구분한다.

```sql
id INTEGER PRIMARY KEY
```

### NOT NULL

NULL 값을 허용하지 않는다.

```sql
name VARCHAR(50) NOT NULL
```

### UNIQUE

같은 값이 중복되는 것을 제한한다.

```sql
email VARCHAR(100) UNIQUE
```

### DEFAULT

값을 입력하지 않았을 때 사용할 기본값을 설정한다.

```sql
created_at TIMESTAMPTZ DEFAULT CURRENT_TIMESTAMP
```

### IDENTITY

값을 직접 입력하지 않아도 번호를 자동으로 생성한다.

# 7. INSERT - 데이터 추가

`INSERT`는 테이블에 새로운 데이터를 추가할 때 사용한다.

## 한 명 추가

```sql
INSERT INTO public.students (name, email, major, grade)
VALUES ('김민지', 'minji@example.com', '컴퓨터공학', 2);
```

문자열은 `' '`로 감싸고 숫자는 따옴표 없이 작성한다.

---

# 8. 여러 데이터를 한 번에 INSERT하기

여러 명을 추가할 때 INSERT 문을 여러 번 작성할 수도 있지만 하나의 `VALUES`에 여러 데이터를 넣을 수도 있다.

```sql
INSERT INTO public.students (name, email, major, grade)
VALUES
    ('이준호', 'junho@example.com', '데이터사이언스', 3),
    ('박서연', 'seoyeon@example.com', '경영학', 1),
    ('최현우', 'hyunwoo@example.com', '컴퓨터공학', 4),
    ('정하늘', 'haneul@example.com', 'AI데이터공학', 2);
```

각 데이터는 `,`로 구분하고 마지막 데이터에만 `;`를 작성한다.

즉,

```text
VALUES
    (첫 번째 데이터),
    (두 번째 데이터),
    (세 번째 데이터);
```

형태로 기억하면 된다.

# 9. SELECT - 데이터 조회

`SELECT`는 저장된 데이터를 조회하는 명령어이다.

## 모든 데이터 조회

```sql
SELECT *
FROM public.students;
```

`*`는 모든 컬럼을 의미한다.

## 필요한 컬럼만 조회

```sql
SELECT id, name, email
FROM public.students;
```

# 10. AS - 별칭 사용

조회 결과의 컬럼 이름을 다르게 표시할 수 있다.

```sql
SELECT
    name AS student_name,
    major AS student_major
FROM public.students;
```

`AS`는 실제 테이블의 컬럼명을 변경하는 것이 아니라 **조회 결과에서 보여주는 이름만 변경한다.**

# 11. WHERE - 조건 조회

`WHERE`를 사용하면 특정 조건을 만족하는 데이터만 조회할 수 있다.

```sql
SELECT *
FROM public.students
WHERE major = '컴퓨터공학';
```

## 비교 연산자

| 연산자 | 의미 |
|---|---|
| `=` | 같다 |
| `<>` | 같지 않다 |
| `>` | 크다 |
| `>=` | 크거나 같다 |
| `<` | 작다 |
| `<=` | 작거나 같다 |

예시:

```sql
SELECT id, name, grade
FROM public.students
WHERE grade >= 3;
```

---

# 12. AND

`AND`는 모든 조건을 만족하는 데이터를 찾는다.

```sql
SELECT *
FROM public.students
WHERE major = '컴퓨터공학'
  AND grade >= 3;
```

위 SQL은

> 전공이 컴퓨터공학이면서 학년이 3 이상

인 학생을 조회한다.

---

# 13. OR

`OR`는 조건 중 하나 이상을 만족하면 된다.

```sql
SELECT *
FROM public.students
WHERE major = '컴퓨터공학'
   OR major = '데이터사이언스';
```

---

# 14. IN

같은 컬럼을 여러 값과 비교할 때 `IN`을 사용하면 편리하다.

```sql
SELECT *
FROM public.students
WHERE major IN ('컴퓨터공학', '데이터사이언스');
```

이는 다음 조건과 비슷한 의미이다.

```sql
WHERE major = '컴퓨터공학'
   OR major = '데이터사이언스'
```

---

# 15. AND와 OR 함께 사용하기

AND와 OR를 같이 사용할 경우 괄호를 사용해서 조건의 범위를 명확하게 하는 것이 좋다.

```sql
SELECT *
FROM public.students
WHERE major = '컴퓨터공학'
  AND (grade = 2 OR grade = 3);
```

# 16. LIKE - 문자열 검색

`LIKE`는 특정 문자열 패턴을 검색할 때 사용한다.

```sql
SELECT *
FROM public.students
WHERE name LIKE '김%';
```

## LIKE 패턴

| 패턴 | 의미 |
|---|---|
| `'김%'` | 김으로 시작 |
| `'%민%'` | 중간에 민 포함 |
| `'%우'` | 우로 끝남 |
| `'김__'` | 김 뒤에 정확히 두 문자 |

### `%`

0개 이상의 문자를 의미한다.

### `_`

정확히 한 문자를 의미한다.

# 17. NULL

`NULL`은 숫자 `0`이나 빈 문자열 `''`과 다르다.

NULL은 **값이 없거나 아직 알 수 없는 상태**를 의미한다.

다음과 같이 작성하면 안 된다.

```sql
WHERE major = NULL;
```

NULL은 일반 값처럼 `=`로 비교하지 않는다.

## NULL 확인

```sql
WHERE major IS NULL;
```

## NULL이 아닌 데이터 확인

```sql
WHERE major IS NOT NULL;
```

# 18. DISTINCT - 중복 제거

조회 결과에서 중복된 값을 제거한다.

```sql
SELECT DISTINCT major
FROM public.students;
```

`DISTINCT`는 원본 데이터를 삭제하는 것이 아니라 **조회 결과에서 중복만 제거한다.**

# 19. ORDER BY - 정렬

조회 결과를 원하는 순서로 정렬할 수 있다.

```sql
SELECT *
FROM public.students
ORDER BY grade ASC;
```

### ASC

오름차순

```text
1 → 2 → 3 → 4
```

### DESC

내림차순

```text
4 → 3 → 2 → 1
```

예시:

```sql
SELECT *
FROM public.students
ORDER BY grade DESC;
```

# 20. LIMIT - 조회 개수 제한

조회할 데이터 개수를 제한할 수 있다.

```sql
SELECT *
FROM public.students
ORDER BY created_at DESC, id DESC
LIMIT 3;
```

위 SQL은 정렬된 결과 중 상위 3개의 데이터만 가져온다.

`LIMIT`을 사용할 때는 원하는 결과 순서를 명확하게 하기 위해 `ORDER BY`를 함께 사용하는 것이 좋다.

# 21. UPDATE - 데이터 수정

`UPDATE`는 기존 데이터를 수정하는 명령어이다.

```sql
UPDATE public.students
SET grade = 4
WHERE email = 'junho@example.com';
```

하지만 UPDATE에서 가장 중요한 것은 `WHERE`이다.

다음 SQL을 실행하면

```sql
UPDATE public.students
SET grade = 1;
```

WHERE가 없기 때문에 모든 학생의 grade가 1로 변경될 수 있다.

---

# 22. UPDATE 안전하게 실행하기

UPDATE를 바로 실행하지 않고 먼저 SELECT로 대상을 확인한다.

## 1단계 - 대상 확인

```sql
SELECT *
FROM public.students
WHERE email = 'junho@example.com';
```

## 2단계 - UPDATE

```sql
UPDATE public.students
SET grade = 4
WHERE email = 'junho@example.com'
RETURNING id, name, grade;
```

## 3단계 - 다시 확인

```sql
SELECT *
FROM public.students
WHERE email = 'junho@example.com';
```

즉,

```text
SELECT로 대상 확인
→ UPDATE 실행
→ 결과 확인
→ SELECT로 다시 확인
```

순서로 진행한다.

---

# 23. DELETE - 데이터 삭제

`DELETE`는 기존 데이터를 삭제한다.

```sql
DELETE FROM public.students
WHERE email = 'seoyeon@example.com';
```

DELETE 역시 `WHERE`가 중요하다.

다음 SQL은

```sql
DELETE FROM public.students;
```

테이블 안에 있는 **모든 데이터를 삭제한다.**

테이블 자체가 삭제되는 것은 아니지만 데이터가 모두 사라진다.

---

# 24. DELETE 안전하게 실행하기

DELETE도 UPDATE와 마찬가지로 먼저 SELECT로 대상을 확인한다.

## 1단계 - 삭제 대상 확인

```sql
SELECT *
FROM public.students
WHERE email = 'seoyeon@example.com';
```

## 2단계 - DELETE

```sql
DELETE FROM public.students
WHERE email = 'seoyeon@example.com'
RETURNING id, name, email;
```

## 3단계 - 다시 조회

```sql
SELECT *
FROM public.students
WHERE email = 'seoyeon@example.com';
```

즉,

```text
SELECT로 대상 확인
→ DELETE 실행
→ 삭제 결과 확인
→ SELECT로 다시 확인
```

순서로 진행한다.

# 25. DBeaver에서 SQL 실행 범위

DBeaver에서는 SQL 전체를 실행할 수도 있고 특정 SQL 문장만 선택해서 실행할 수도 있다.

예를 들어 다음 SQL이 있다고 가정한다.

```sql
SELECT 'A' AS step;
SELECT 'B' AS step;
SELECT 'C' AS step;
```

한 문장만 실행하면 해당 SQL만 실행된다.

```sql
SELECT 'A' AS step;
```

결과:

```text
A
```

A와 B만 선택해서 실행하면

```sql
SELECT 'A' AS step;
SELECT 'B' AS step;
```

선택한 SQL만 실행된다.

전체 스크립트를 실행하면 A, B, C가 모두 실행된다.

따라서 SQL 실행 전에는 **현재 어떤 문장 또는 영역을 선택했는지 확인해야 한다.**

# 28. SQL 오류가 발생했을 때 확인 순서

SQL 오류가 발생했다고 해서 전체 코드를 바로 다시 작성하지 않는다.

다음 순서로 확인한다.

```text
오류 메시지 확인
↓
문제가 발생한 SQL 위치 확인
↓
테이블명 / 컬럼명 확인
↓
데이터 타입 확인
↓
문자열 따옴표 확인
↓
쉼표와 괄호 확인
↓
SQL 수정
↓
다시 실행
```

예를 들어 다음처럼 SQL을 붙여서 작성하면 오류가 발생할 수 있다.

```sql
nameVARCHAR(50)NOTNULL
```

올바르게 작성하면

```sql
name VARCHAR(50) NOT NULL
```

이 된다.

오류 메시지를 확인하면 문제가 발생한 위치를 찾는 데 도움이 된다.

# 29. AI가 만든 SQL 사용할 때 주의할 점

AI가 SQL을 만들어주더라도 바로 실행하지 않는다.

특히 다음과 같은 SQL은 주의해야 한다.

```sql
DELETE FROM public.students
WHERE major = '컴퓨터공학';
```

먼저 같은 조건으로 SELECT를 실행한다.

```sql
SELECT *
FROM public.students
WHERE major = '컴퓨터공학';
```

조회 결과를 보고 정말 해당 데이터들을 삭제하는 것이 맞는지 확인한 뒤 실행해야 한다.

AI가 작성한 코드도 최종적으로 실행하고 검증하는 것은 개발자의 역할이다.

# 30. 자주 할 수 있는 실수

SQL을 사용할 때 다음 실수를 주의한다.

- 현재 연결된 DB를 확인하지 않음
- SQL 전체를 실수로 실행함
- 문자열에 작은따옴표를 사용하지 않음
- `NULL`을 `= NULL`로 비교함
- AND와 OR 조건의 범위를 확인하지 않음
- ORDER BY 없이 결과 순서를 예상함
- WHERE 없이 UPDATE 실행
- WHERE 없이 DELETE 실행
- INSERT에서 값과 컬럼 순서를 잘못 작성함
- 쉼표 `,`를 빼먹음
- 테이블명이나 컬럼명을 잘못 작성함
- AI가 만든 SQL을 확인하지 않고 바로 실행함
