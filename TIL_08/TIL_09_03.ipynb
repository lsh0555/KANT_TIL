## Chapter02 -  데이터와 DBMS의 기본 개념

### 데이터(Data) : 관찰하거나 기록한 값이나 사실

예시:
- 학생 이름
- 학번
- 강의 제목
- 수강신청일
- 신청 상태

### 데이터베이스(Database) : 관련된 데이터를 일정한 구조로 저장하고 관리하는 공간

예:
`ai_database`

### DBMS(Database Management System) : 데이터베이스를 생성하고 관리하며 사용할 수 있도록 해주는 소프트웨어

대표적인 DBMS:
- PostgreSQL
- MySQL
- MariaDB
- Oracle Database
- SQL Server
- SQLite


### DBeaver : 데이터베이스에 접속해서 SQL을 작성하고 결과를 확인하는 클라이언트 프로그램

### PostgreSQL : 실제로 데이터를 저장하고 SQL을 실행하는 DBMS

### SQL : DBMS에게 데이터 조회, 추가, 수정, 삭제 등의 작업을 요청하는 명령 언어

### 실행 과정

사용자
↓
DBeaver에서 SQL 작성
↓
PostgreSQL에 SQL 전달
↓
PostgreSQL이 SQL 실행
↓
결과 또는 오류 반환
↓
DBeaver가 결과를 화면에 보여줌


## 3. PostgreSQL의 기본 구조

PostgreSQL 서버
└── 데이터베이스
    └── 스키마
        └── 테이블
            └── 행 / 열

예:
PostgreSQL
└── ai_database
    └── public
        └── students

### 스키마(Schema) : 데이터베이스 안에서 테이블 등의 객체를 구분하고 관리하는 공간

테이블을 정확하게 지정할 때는 다음과 같이 작성할 수 있다.

`public.students`



### 테이블(Table) : 같은 성격의 데이터를 행과 열 형태로 저장하는 구조

예:
| id | name | email |
|---|---|---|
| 1 | 민수 | minsu@test.com |
| 2 | 지영 | jiyoung@test.com |

### 행(Row) : 하나의 데이터 기록을 의미
* 테이블에서 한 행이 무엇을 의미하는지 명확하게 정하는 것 중요

`한 행 = 학생 한 명`

### 열(Column) : 데이터를 구성하는 항목

예:
- id
- name
- email


### 테이블 : 데이터가 실제로 저장되어 있는 구조

### 조회 결과(Result Set) : SQL을 실행해서 만들어진 결과

예:
SELECT *
FROM students;

화면에 표 형태로 결과가 나오더라도 이것은 원본 테이블 자체가 아니라 SQL을 실행해서 얻은 조회 결과이다.


## 6. 데이터 타입 : 각 열에는 저장할 데이터의 종류를 나타내는 데이터 타입이 있다.

예:

- 정수 → id, 수량
- 문자열 → 이름, 이메일, 학번
- 날짜/시간 → 가입일, 신청일
- Boolean → 활성 여부, 완료 여부

### 주의

숫자로 생겼다고 무조건 숫자 타입을 사용하는 것은 아니다.

예를 들어 학번은

`20260001`

처럼 숫자로 보이지만 계산을 위한 값이 아니기 때문에 문자열로 관리할 수도 있다.

따라서 데이터의 모양보다 **업무에서 어떻게 사용하는 값인지**를 생각해야 한다.

## 7. 내부 식별자와 업무 식별자

### 내부 식별자

DB 내부에서 데이터를 안정적으로 구분하기 위한 값

예:

`students.id`

### 업무 식별자

실제 업무에서 사람이나 대상을 구분하기 위해 사용하는 값

예:
- 학번
- 회원번호
- 주문번호

예를 들어

id = 1
student_number = "20260001"

이라면

`id`는 DB 내부에서 학생을 구분하고,
`student_number`는 학교 업무에서 학생을 구분한다.

## 8. 기본키(PK)

기본키(Primary Key)는 테이블에서 각각의 행을 고유하게 구분하는 값이다.

예:

| id | name |
|---|---|
| 1 | 김민지 |
| 2 | 김민지 |

이름이 같더라도 id가 다르기 때문에 서로 다른 학생으로 구분할 수 있다.

### 기본키 특징

- 값이 중복되지 않는다.
- NULL이 될 수 없다.
- 각 행을 고유하게 구분한다.

---

## 9. 외래키(FK)

외래키(Foreign Key)는 다른 테이블의 데이터를 참조하여 테이블 사이의 관계를 연결한다.

예:

students

| id | name |
|---|---|
| 1 | 민수 |

questions

| id | student_id | question |
|---|---|---|
| 1 | 1 | JOIN 질문 |
| 2 | 1 | NULL 질문 |

여기서

`questions.student_id`

는

`students.id`

를 참조한다.

### 핵심

PK → 자기 테이블의 행을 구분

FK → 다른 데이터와의 연결을 표현

외래키는 1:N 관계에서 같은 값이 여러 번 등장할 수도 있다.

---

## 10. 데이터의 관계

### 1:1

한 데이터가 다른 데이터 하나와 연결

예:
사용자 ↔ 상세 프로필

### 1:N

한 데이터가 여러 데이터와 연결

예:
학생 1명 → 질문 여러 개

### N:M

양쪽 데이터가 서로 여러 개 연결

예:
학생 여러 명 ↔ 강의 여러 개

N:M 관계는 일반적으로 중간에 연결 테이블을 만들어 관리한다.

학생
1
↓
N
수강신청
N
↓
1
강의

## 11. DBMS의 제약조건

DBMS는 잘못된 데이터가 저장되는 것을 줄이기 위해 제약조건을 사용할 수 있다.

대표적인 제약조건:

- PRIMARY KEY
- FOREIGN KEY
- NOT NULL
- UNIQUE
- CHECK

예:

- 학생 이름은 NULL이 될 수 없다.
- 이메일은 중복될 수 없다.
- 존재하지 않는 학생의 수강신청을 저장할 수 없다.
- 가격은 0보다 작을 수 없다.

## 12. SQL과 CRUD

CRUD는 데이터를 다룰 때 사용하는 기본적인 네 가지 작업이다.

| CRUD | 의미 | SQL |
|---|---|---|
| Create | 데이터 추가 | INSERT |
| Read | 데이터 조회 | SELECT |
| Update | 데이터 수정 | UPDATE |
| Delete | 데이터 삭제 | DELETE |

### 주의

CRUD의 `Create`와 SQL의 `CREATE TABLE`은 의미가 다르다.

`INSERT`
→ 새로운 데이터 행 추가

`CREATE TABLE`
→ 새로운 테이블 구조 생성

## 13. ORDER BY가 중요한 이유

SQL에서 데이터를 조회했다고 해서 행의 순서가 항상 일정한 것은 아니다.

정렬된 결과가 필요하다면 `ORDER BY`를 사용해야 한다.

예:

SELECT *
FROM students
ORDER BY id;

따라서 `ORDER BY`가 없다면 조회 결과의 순서를 당연하게 가정하면 안 된다.

## 14. AI가 만든 데이터 구조 검토하기

AI가 SQL이나 테이블 구조를 만들어 주더라도 바로 사용하면 안 된다.

다음 내용을 확인해야 한다.

1. 이 테이블의 한 행은 무엇을 의미하는가?
2. 각 행을 구분할 기본키가 있는가?
3. 다른 테이블을 참조해야 하는 값은 무엇인가?
4. 서로 다른 성격의 데이터가 한 테이블에 섞여 있지 않은가?
5. 데이터 타입과 필수 여부가 데이터의 의미와 맞는가?

SQL이 실행된다는 것과 데이터 구조가 올바르다는 것은 서로 다른 문제이다.

# 오늘의 핵심 정리

### 1. DBeaver와 PostgreSQL은 다르다.

DBeaver는 SQL을 작성하고 결과를 확인하는 클라이언트이고,
PostgreSQL은 실제 데이터를 저장하고 SQL을 실행하는 DBMS이다.

### 2. PK와 FK의 역할이 다르다.

PK는 자신의 테이블에서 데이터를 고유하게 구분하고,
FK는 다른 테이블의 데이터와 연결하는 역할을 한다.

### 3. SQL이 실행된다고 데이터 구조가 올바른 것은 아니다.

SQL이 오류 없이 실행되더라도 한 행의 의미, PK/FK, 데이터 타입, 업무 요구사항 등이 제대로 반영되었는지 확인해야 한다.
